Exercise 3: Generative AI — Prompt Engineering for Video Synthesis
1. Selected AI Video Generation Tool
Tool Name: Google Veo (Google DeepMind)

Website: https://deepmind.google/models/veo/

2. Objective
To generate a 3-second AI-generated video using three images from Exercise 2:

Image 1: First source image.
Image 2: Second source image.
Image 3: Blended image created using OpenCV.
3. Exact Prompt Text
"Create a professional 3-second cinematic video using the three provided reference images. Begin by displaying the first source image clearly for a brief moment. Smoothly transition into the second source image using an elegant crossfade effect. Finally, blend both scenes into the third blended image, creating a seamless visual transformation. Maintain the original colors, objects, textures, and composition of all three images. Use smooth motion, realistic visual effects, and professional cinematic quality. Ensure that the final frame clearly displays the blended image."

4. Camera and Transition Instructions
Duration: 3 seconds.
Camera Movement: Slow and smooth zoom-in effect.
Transition 1: Smooth crossfade from Image 1 to Image 2.
Transition 2: Gradual blending transition from Image 2 to the blended Image 3.
Visual Effects: Smooth motion and natural lighting.
Final Frame: Clearly display the blended image.
Video Quality: High-quality cinematic output.
5. Expected Output
A 3-second AI-generated video demonstrating a smooth visual transition between the two original source images and their blended image.
