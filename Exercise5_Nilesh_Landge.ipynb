```python
import gradio as gr
from transformers import pipeline

# Load AI language model
print("Loading AI Model...")

chatbot = pipeline(
    "text-generation",
    model="google/flan-t5-small"
)

# Define chatbot response function
def respond(message, history):

    if not message.strip():
        return "Please enter a question."

    result = chatbot(
        message,
        max_new_tokens=150,
        do_sample=True,
        temperature=0.7
    )

    return result[0]["generated_text"]


# Create Gradio chatbot interface
demo = gr.ChatInterface(
    fn=respond,
    title="🤖 My AI Assistant Chatbot",
    description="""
    Welcome to My AI Chatbot!

    This chatbot uses Artificial Intelligence
    to answer your questions.

    Ask questions about Python, Artificial
    Intelligence, Machine Learning and Technology.
    """,
    examples=[
        "What is Artificial Intelligence?",
        "Explain Python programming.",
        "What is Machine Learning?",
        "How does a chatbot work?"
    ],
    chatbot=gr.Chatbot(
        height=400,
        placeholder="Welcome! Ask me anything..."
    )
)

# Launch chatbot
demo.launch()
```
