```python

# STEP 1: Import required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# STEP 2: Generate random numerical data

np.random.seed(42)  # For reproducible results

data = {
    'f1': np.random.randint(1, 50, 20),
    'f2': np.random.randint(1, 50, 20),
    'f3': np.random.randint(1, 50, 20),
    'f4': np.random.randint(1, 50, 20)
}
```

```python
# STEP 3: Create Pandas DataFrame

df = pd.DataFrame(data)

print("GENERATED DATASET:")
display(df)
```

```python
# STEP 4: Generate Scatter Plot between f2 and f3

plt.figure(figsize=(8, 5))

plt.scatter(df['f2'], df['f3'], color='pink', marker='o')

plt.xlabel('Feature f2')
plt.ylabel('Feature f3')
plt.title('Scatter Plot: Relationship between f2 and f3')

plt.grid(True)
plt.show()
```

```python
# STEP 5: Calculate Row-wise Arithmetic Mean

df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

print("DATASET WITH NEW COLUMN f5:")
display(df)
```

```python
# STEP 6: Export DataFrame to Excel

file_name = 'Exercise1_Report.xlsx'

df.to_excel(file_name, index=False)

print("Excel file exported successfully!")
print("File Name:", file_name)
```

```python
# STEP 7: Display Statistical Summary

print("\nSTATISTICAL SUMMARY:")
display(df.describe())
```
